## Summarize doc chain

```python
from pypdf import PdfReader
from langchain_core.documents import Document
from langchain_core.prompts import ChatPromptTemplate
```

```python
from langchain.chat_models import init_chat_model
llm = init_chat_model(
    "claude-sonnet-4-6",
    model_provider="anthropic",
    temperature=1
)
```

```python
!wget https://www.cheat-sheets.org/saved-copy/The%20One%20Page%20Linux%20Manual.pdf
```

```python
reader = PdfReader("./The One Page Linux Manual.pdf")
# print(reader.pages)
text = "\n".join(
    page.extract_text() or ""
    for page in reader.pages
)
# print(text)
```

```python
prompt = ChatPromptTemplate.from_template(
    "Summarize the following document:\n\n{text}"
)
```

```python
chain = prompt | llm

response = chain.invoke({"text": text})

print(response.content)
```

## QA Chain

```python
from langchain_core.prompts import PromptTemplate
from langchain.chat_models import init_chat_model

llm = init_chat_model(
    "claude-sonnet-4-6",
    model_provider="anthropic",
    temperature=0
)
prompt = PromptTemplate(template="Question: {question}\nAnswer:", input_variables=["question"])
chain = prompt | llm
```

```python
response = chain.invoke(" What is the meaning of life?")
print(response.content)
```
